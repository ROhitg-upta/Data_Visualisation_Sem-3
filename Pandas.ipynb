```python
import numpy as np
import pandas as pd
```

```python
data = {'Company': ['GOOG', 'GOOG', 'MSFT', 'MSFT', 'FB', 'FB'],
        'Person': ['Sam', 'Charlie', 'Amy', 'Vanessa', 'Carl', 'Sarah'],
        'sales': [200, 120, 340, 124, 243, 350]}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
import numpy as np
import pandas as pd
byComp = df.groupby('Company')
```

```python
byComp['sales'].mean()
```

```python
byComp.sum().loc['FB']
```

```python
byComp.sum().loc['FB']
```

  Joining

```python
import numpy as np
import pandas as pd

left = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3'],
                     'A': ['A0', 'A1', 'A2', 'A3']},
                     index = ['B0', 'B1', 'B2', 'B3'])

right = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3'],
                      'B': ['B0', 'B1', 'B2', 'B3']},
                      index = ['B0', 'B1', 'B2', 'B3'])
```

```python
left.join(right, how='outer', lsuffix='_left', rsuffix='_right')
```
