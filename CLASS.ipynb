```python
import matplotlib.pyplot as plt
students = ["Vikky", "Rahul", "Aman", "Neha"]
marks = [85,90,78,95]
plt.bar(students, marks, width = 0.6)
plt.title("Student Marks")
plt.xlabel("students")
plt.ylabel("marks")
# plt.barh(students, marks)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D"]

python = [80, 70, 90, 60]
java = [70, 75, 85, 80]

x = np.arange(len(students))

width = 0.35

plt.bar(x - width/2, python, width, label="Python")
plt.bar(x + width/2, java, width, label="Java")

plt.xticks(x, students)
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["AI", "Python", "Java", "SQL"]

hours = [35, 25, 20, 20]

explode = [0.1, 0, 0, 0]

mycolors = ["Red", "hotpink", "b", "#4CAF50"]

plt.pie(
    hours,
    labels=subjects,
    autopct="%1.1f%%",
    explode=explode,
    shadow=True,
    colors=mycolors
)

plt.title("Study Time")
plt.legend()
plt.show()
```

Scatter Plot

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [35, 42, 50, 58, 70, 78]

plt.scatter(hours, marks)

plt.xlabel("Hours Studied")
plt.ylabel("Marks")
plt.title("Hours Studied vs Marks")

plt.show()
```
